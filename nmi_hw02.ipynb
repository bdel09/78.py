### b) (0 pts) fixed-point

<font color = orange>not needed for homework; for continuity of methods.</font>

#### algorithm

$x_{i+1} = x_{i} - f(x_{i})$


#### code, with f(x) as argument

````python
# fixed point
def f_fpi(ag,ax0,atol,aimax):
  rc = 0
  isteps = 0
  x = ax0
  gx = ag(x)
  while (abs(gx - x) > atol) and (isteps < aimax):
    x = gx
    gx = ag(x)
    isteps += 1 # common shortcut for isteps = isteps + 1
  rc = x
  return rc,isteps
````

### c) (10 pts) newton-raphson

#### algorithm

newton-raphson: $x_{new} = x_{old} - \frac{f(x_{old})}{f'(x_{old})}$


```
icount = 0
fx = f(x_old)
while (abs(fx) > epsilon) and (icount <= imax):
  x_new = x_old - f(x_old)/f'(x_old)
  fx = f(x_new)
  x_old = x_new
  icount = icount + 1
return x_old
```
